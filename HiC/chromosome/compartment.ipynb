```python
import os

import pandas as pd
```

```python
dir = "/data/data/Turner/HiCData/ZWJFamilyHiCAfterMapping/outPut/step04_hicpro2juicebox/hicfiles/compartment"

columns_name = ["chrom", "start", "end", "score"]
compartment = {
    "ZF_naive_X": pd.read_csv(os.path.join(dir, "F-naive/F-naive_pc1_ori.bdg"), sep="\t", names=columns_name),
    "ZM_naive_X": pd.read_csv(os.path.join(dir, "M-naive/M-naive_pc1_ori.bdg"), sep="\t", names=columns_name),
    "Z_naive_X": pd.read_csv(os.path.join(dir, "Z-naive/Z-naive_pc1_ori.bdg"), sep="\t", names=columns_name),
    "ZF_IgDn_X": pd.read_csv(os.path.join(dir, "F-IgD-/F-IgD-_pc1_ori.bdg"), sep="\t", names=columns_name),
    "ZM_IgDn_X": pd.read_csv(os.path.join(dir, "M-IgD-/M-IgD-_pc1_ori.bdg"), sep="\t", names=columns_name),
    "Z_IgDn_X": pd.read_csv(os.path.join(dir, "Z-IgD-/Z-IgD-_pc1_ori.bdg"), sep="\t", names=columns_name),
}

pdb_path = {
    'Z_naive_X': './Z_naive_X/concat_1735275745875.pdb',
    'Z_IgDn_X': './Z_IgDn_X/concat_1735275871942.pdb',
    'ZF_naive_X': './ZF_naive_X/concat_1735276762067.pdb',
    'ZF_IgDn_X': './ZF_IgDn_X/concat_1735276896226.pdb',
    'ZM_naive_X': './ZM_naive_X/concat_1735276909833.pdb',
    'ZM_IgDn_X': './ZM_IgDn_X/concat_1735277362861.pdb',
}
```

```python
def modified_pdb(pdb_path, B_indexs, key):
    with open(pdb_path[key], 'r') as file:
        lines = file.readlines()
    with open("./compartment/{}.pdb".format(key), 'w') as f:
        for _index in B_indexs:
            lines[_index] = lines[_index].replace("CA", "CB")
        for line in lines:
            f.write(line)

step_len = 500000
for key, value in compartment.items():
    B_indexs = []
    for i in range(0, value["end"].max(), step_len):
        _value = value[(value["start"] >= i) & (value["end"] <= i + step_len)]
        _score = _value["score"].mean()
        if _score < 0:
            B_indexs.append((i//step_len)+1)
    modified_pdb(pdb_path, B_indexs, key)
```
